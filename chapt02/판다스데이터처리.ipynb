## 판다스 데이터 처리

### 데이터 정제

- 결측값(치)와 이상치를 제거하고 분석가능한 깨끗한 데이터로 수정하는 작업
    - 결측값 : NULL, None, Nan 등 입력되지 않은 필요한 데이터. 나이 컬럼에 값 없음 
    - 이상치 : 컬럼 도메인에 알맞지 않은 데이터. 점수 999점  
    - 중복값 : 중복된 데이터. 같은 학생이 두 번 등록
    - 타입오류 : 일치하지 않는 데이터 타입으로 입력된 값. 점수에 문자열로 저장. '구십구점'
    - 불필요한 공백 : 필요없는 스페이스키 입력 '  홍길동  '
    - 도메인 오류 : 표현방식 일치. 'M', 'Male', '남자', '남성'...
- 위 문제들을 해결하지 않으면 통계(계산)를 낼 수 없음

### 데이터 전처리

- 데이터 정제를 하는 과정, 분석 프로젝트의 40~50% 차지 
- AI 시대에서는 정제 과정이 많이 축소될 것으로 예상
- 사람의 검수 반드시 필요. 시간 대부분 여기에서 소요
